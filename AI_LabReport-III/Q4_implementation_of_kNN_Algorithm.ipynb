# 4). Implement k-Nearest Neighbors(kNN) algorithm

```python
# Implementation of kNN algorithm

import numpy as np
from collections import Counter

def knn(X_train, y_train, x_test, k):
    if k > len(X_train):
        raise ValueError("k cannot be greater than number of samples")

    distances = []
    for x, label in zip(X_train, y_train):
        dist = np.linalg.norm(x - x_test)
        distances.append((dist, label))

    distances.sort(key=lambda x: x[0])
    k_labels = [label for _, label in distances[:k]]

    return Counter(k_labels).most_common(1)[0][0]

# Data
X_train = np.array([[1,2], [2,3], [3,3], [6,5]])
y_train = np.array([0, 0, 1, 1])
x_test = np.array([2,2])

print("Predicted Class:", knn(X_train, y_train, x_test, k=3))
```
