# 3)To design and train a perceptron for NOT gate.

```python
import numpy as np

# NOT gate data
X = np.array([[0], [1]])
y = np.array([1, 0])

# Initialize
w = np.zeros(1)
b = 0
lr = 0.1

# Step function
def step(z):
    return 1 if z >= 0 else 0

# Training
for _ in range(10):
    for i in range(len(X)):
        z = np.dot(X[i], w) + b
        y_pred = step(z)
        error = y[i] - y_pred
        w += lr * error * X[i]
        b += lr * error

print("Weight:", w)
print("Bias:", b)

# Testing
print("NOT Gate Output:")
for x in X:
    print(x[0], "->", step(np.dot(x, w) + b))
```
