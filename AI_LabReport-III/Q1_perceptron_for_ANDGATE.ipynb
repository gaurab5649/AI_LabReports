# 1). To Design and Train a perceptron for AND Gate.

```python
import numpy as np

# AND gate dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
Y = np.array([0, 0, 0, 1])

# Initialize weights and bias
w = np.zeros(2)
b = 0
lr = 0.1

# Step activation function
def step(z):
    return 1 if z >= 0 else 0

# Training
for epoch in range(10):
    for x, y in zip(X, Y):
        z = np.dot(w, x) + b
        y_hat = step(z)
        error = y - y_hat
        w += lr * error * x
        b += lr * error

# Testing
print("Weights:", w)
print("Bias:", b)

for x in X:
    print(x, "->", step(np.dot(w, x) + b))
```
