# 2). To design and train a perceptron training for X-OR gate.
---
## DESCRIPTION
Single-Layer Perceptron FAILS because:
A single-layer perceptron:Can only learn linearly separable problems & Uses one straight decision boundary

Let us look at XOR data points:

(0,1) and (1,0) → class 1

(0,0) and (1,1) → class 0

There is NO single straight line that separates these two classes.

```python
import numpy as np

X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 1, 1, 0])

w = np.zeros(2)
b = 0
lr = 0.1

for _ in range(10):
    for i in range(len(X)):
        y_pred = 1 if np.dot(X[i], w) + b >= 0 else 0
        w += lr * (y[i] - y_pred) * X[i]
        b += lr * (y[i] - y_pred)
print("We cannot design and train a single-layer perceptron for an XOR gate\n")
print("Failed XOR Gate Output :")
for x in X:
    print(x, "->", 1 if np.dot(x, w) + b >= 0 else 0)
```
