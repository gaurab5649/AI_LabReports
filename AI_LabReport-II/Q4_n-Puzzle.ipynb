# Q4. Implement n-Puzzle search problem

```python
from collections import deque

class NPuzzle:
    def __init__(self, initial_state, goal_state):
        self.initial_state = initial_state  # tuple of tuples
        self.goal_state = goal_state
        self.size = len(initial_state)  # e.g., 3 for 3x3 puzzle

    def goalTest(self, state):
        return state == self.goal_state

    def find_blank(self, state):
        """Return (row, col) of the blank (0)"""
        for i in range(self.size):
            for j in range(self.size):
                if state[i][j] == 0:
                    return i, j

    def successor(self, state):
        """Generate all valid moves from the current state"""
        row, col = self.find_blank(state)
        successors = []

        # Directions: up, down, left, right
        directions = [(-1,0), (1,0), (0,-1), (0,1)]
        for dr, dc in directions:
            new_r, new_c = row + dr, col + dc
            if 0 <= new_r < self.size and 0 <= new_c < self.size:
                # Swap blank with adjacent tile
                new_state = [list(r) for r in state]  # deep copy
                new_state[row][col], new_state[new_r][new_c] = new_state[new_r][new_c], new_state[row][col]
                successors.append(tuple(tuple(r) for r in new_state))
        return successors

    def BFS(self):
        """Breadth First Search"""
        queue = deque([(self.initial_state, None)])
        CLOSED = {}

        while queue:
            state, parent = queue.popleft()
            if state not in CLOSED:
                CLOSED[state] = parent
                if self.goalTest(state):
                    return self.generate_path(CLOSED, state)
                for child in self.successor(state):
                    queue.append((child, state))
        return None

    def generate_path(self, CLOSED, state):
        """Reconstruct path from initial to goal"""
        path = []
        while state is not None:
            path.append(state)
            state = CLOSED[state]
        path.reverse()
        return path


# let us assume an example

# 8-Puzzle Example (3x3)
initial_state = (
    (1, 2, 3),
    (4, 0, 6),
    (7, 5, 8)
)

goal_state = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

puzzle = NPuzzle(initial_state, goal_state)
solution_path = puzzle.BFS()

if solution_path:
    print("BFS Solution Path:")
    for step, state in enumerate(solution_path):
        print(f"\nStep {step}:")
        for row in state:
            print(row)
else:
    print("No solution found.")
```
