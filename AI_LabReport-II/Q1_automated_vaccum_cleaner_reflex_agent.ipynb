# Q1. Implement automated vacuum cleaner reflex agent.

```python
#Implementation of automated vacuum cleaner reflex agent

# Defining the Reflex Agent
class VacuumCleanerAgent:
    def __init__(self):
        self.location = 'A'

    def perceive(self, environment):
        return self.location, environment[self.location]

    def act(self, percept):
        location, status = percept
        if status == 'Dirty':
            return 'SUCK'
        elif location == 'A':
            return 'RIGHT'
        else:
            return 'LEFT'

    def execute(self, action, environment):
        if action == 'SUCK':
            environment[self.location] = 'Clean'
        elif action == 'RIGHT':
            self.location = 'B'
        elif action == 'LEFT':
            self.location = 'A'
```

```python
# Defining Environment and Create Agent
environment = {'A': 'Dirty', 'B': 'Dirty'}
agent = VacuumCleanerAgent()

# Run the Agent Simulation
for step in range(5):
    percept = agent.perceive(environment)
    action = agent.act(percept)
    agent.execute(action, environment)

    print(f"Step {step + 1}")
    print(f"Percept: Location = {percept[0]}, Status = {percept[1]}")
    print(f"Action: {action}")
    print(f"Environment: A = {environment['A']}, B = {environment['B']}")
    print("-" * 40)

print("\nFinal Environment Status:")
for loc in environment:
    print(f"Location {loc}: {environment[loc]}")
```
