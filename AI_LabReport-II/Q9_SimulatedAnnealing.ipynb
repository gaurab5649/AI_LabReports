# Q9. WAP to demonstrate the effect of temperature on the probability of choosing an inferior node by selecting an appropriate temperature schedule.


```python
import math
import random

# Example cost function: f(x) = (x-3)^2 (min at x=3)
def cost(x):
    return (x - 3) ** 2

# Simulated Annealing
def simulated_annealing(start, initial_temp, alpha, iterations):
    current = start
    current_cost = cost(current)
    path = [(current, current_cost, initial_temp)]

    T = initial_temp

    for i in range(iterations):
        # Generate neighbor (small random step)
        neighbor = current + random.uniform(-1, 1)
        neighbor_cost = cost(neighbor)

        delta = neighbor_cost - current_cost

        # Accept if better, or probabilistically if worse
        if delta < 0 or random.random() < math.exp(-delta / T):
            current = neighbor
            current_cost = neighbor_cost

        path.append((current, current_cost, T))

        # Decrease temperature (exponential schedule)
        T *= alpha

    return path


# Driver Code
start = random.uniform(0, 6)   # random initial state
initial_temp = 10
alpha = 0.9                     # cooling rate
iterations = 20

result_path = simulated_annealing(start, initial_temp, alpha, iterations)

print("Step\tState\tCost\tTemperature")
for i, (state, c, temp) in enumerate(result_path):
    print(f"{i}\t{state:.4f}\t{c:.4f}\t{temp:.4f}")
```
