# Q8. Write a program to implement the steepest ascent hill climbing for the 8-puzzleproblem. Develop appropriate heuristic functions.


```python
import copy

# Goal state
GOAL_STATE = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

# Moves: up, down, left, right
MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def manhattan_distance(state):
    """Heuristic: Sum of Manhattan distances of tiles from goal positions."""
    distance = 0
    for i, tile in enumerate(state):
        if tile != 0:
            goal_index = GOAL_STATE.index(tile)
            x1, y1 = divmod(i, 3)
            x2, y2 = divmod(goal_index, 3)
            distance += abs(x1 - x2) + abs(y1 - y2)
    return distance


def get_neighbors(state):
    """Generate all valid successor states by moving the blank."""
    neighbors = []
    zero_index = state.index(0)
    x, y = divmod(zero_index, 3)

    for dx, dy in MOVES:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_index = nx * 3 + ny
            new_state = list(state)
            new_state[zero_index], new_state[new_index] = new_state[new_index], new_state[zero_index]
            neighbors.append(tuple(new_state))

    return neighbors


def steepest_ascent_hill_climb(start_state):
    """Steepest Ascent Hill Climbing for 8-Puzzle."""
    current = start_state
    path = [current]

    while True:
        neighbors = get_neighbors(current)
        # Choose neighbor with lowest heuristic
        next_state = min(neighbors, key=manhattan_distance)
        
        if manhattan_distance(next_state) >= manhattan_distance(current):
            # No improvement, local maximum reached
            break

        current = next_state
        path.append(current)

        if current == GOAL_STATE:
            break

    return path


#  Driver Code
start_state = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

solution_path = steepest_ascent_hill_climb(start_state)

print("Solution Path with Heuristic Values:")
for step in solution_path:
    h = manhattan_distance(step)
    print(f"State: {step}, Heuristic: {h}")

if solution_path[-1] != GOAL_STATE:
    print("\nNote: Algorithm stopped at a local maximum; solution may not be optimal or complete.")
```
