## Q5. WAP in Python to calculate the heuristic value of the states for the Blocks World Problem.

### Global Heuristic
The heuristic value **e(p)** is calculated as follows:

- For each block that has the correct support structure, give +1 to every block in the support structure.
- For each block that has a wrong support structure, give −1 to every block in the support structure.


```python

# Blocks World Heuristic Calculation


def heuristic(state, goal):
    """
    Calculate global heuristic e(p) for a given Blocks World state.
    
    Args:
        state: Current state as a list of stacks (list of lists, bottom-to-top)
        goal: Goal state as a list of stacks (list of lists, bottom-to-top)
    
    Returns:
        Heuristic value (higher means closer to goal)
    """
    h = 0
    for stack_index, stack in enumerate(state):
        for level, block in enumerate(stack):
            if stack_index < len(goal) and level < len(goal[stack_index]):
                if block == goal[stack_index][level]:
                    # Correct support structure → add +1 to every block in support
                    h += level + 1
                else:
                    # Wrong support structure → subtract -1 for each block in support
                    h -= level + 1
            else:
                # Extra block in stack → count as wrong support
                h -= level + 1
    return h

# Let us consider an example 

# Start state
start_state = [
    ['B', 'A'],
    ['C'],
    ['D']
]

# Goal state
goal_state = [
    ['A', 'B'],
    ['C', 'D'],
    []
]

# Intermediate state
intermediate_state = [
    ['A', 'B'],
    ['D', 'C'],
    []
]


# Calculating heuristic values
print("Start State:", start_state)
print("Heuristic:", heuristic(start_state, goal_state), "\n")

print("Intermediate State:", intermediate_state)
print("Heuristic:", heuristic(intermediate_state, goal_state), "\n")

print("Goal State:", goal_state)
print("Heuristic:", heuristic(goal_state, goal_state))
```
