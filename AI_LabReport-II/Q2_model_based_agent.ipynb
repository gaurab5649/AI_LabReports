## Why  Model-Based Agent is more efficient 

1. **Memory of Environment**  
   - Unlike a simple reflex agent, the model-based agent remembers which locations are clean or dirty, which prevents the agent from revisiting already-clean locations unnecessarily.

2. **Reduces Redundant Actions**  
   - By using the internal model, the agent moves only to locations that still require cleaning, minimizing wasted movements and increases overall efficiency.

3. **Ability to Stop When Done**  
   - The agent can detect when all locations are clean (`NoOp`) and terminate the simulation early.  
   - Reflex agents continue moving even when cleaning is complete.

4. **Scalability**  
   - For larger environments, remembering past actions allows the agent to plan an efficient path.  
   - Reflex agents cannot scale well because they only react to the current percept.

5. **Intelligent Decision-Making**  
   - Chooses actions based on **current percept + internal model**, not just immediate stimulus, demonstrating a more advanced level of autonomy.




```python
# Implementation of an model-based agent

# Defining Model-Based Vacuum Cleaner Agent

class ModelBasedVacuumAgent:
    def __init__(self):
        self.location = 'A'
        self.model = {'A': 'Unknown', 'B': 'Unknown'}

    def perceive(self, environment):
        """Returns current location and its status"""
        return self.location, environment[self.location]

    def update_model(self, percept):
        location, status = percept
        self.model[location] = status

    def decide(self):
        """Decide next action based on model"""
        if self.model[self.location] == 'Dirty':
            return 'Suck'
        # Move to a dirty location if exists
        for loc, status in self.model.items():
            if status == 'Dirty' and loc != self.location:
                return 'Move'
        # Stop if everything is clean
        return 'NoOp'

    def act(self, action, environment):
        if action == 'Suck':
            environment[self.location] = 'Clean'
            self.model[self.location] = 'Clean'
        elif action == 'Move':
            self.location = 'B' if self.location == 'A' else 'A'
```

```python
# Environment definition: The environment consists of two locations: A and B, both initially dirty.

environment = {'A': 'Dirty', 'B': 'Dirty'}
agent = ModelBasedVacuumAgent()

# Simulation loop: The agent runs until all locations are clean or a maximum of 10 steps.

for step in range(10):
    percept = agent.perceive(environment)
    agent.update_model(percept)
    action = agent.decide()

    # Print nicely formatted output
    print(f"Step {step + 1}")
    print(f"Location: {agent.location}")
    print(f"Percept: Location = {percept[0]}, Status = {percept[1]}")
    print(f"Action: {action}")
    print(f"Model: {agent.model}")
    print(f"Environment: A = {environment['A']}, B = {environment['B']}")
    print("-" * 50)

    if action == 'NoOp':
        print("All locations are clean. Simulation stopped.")
        break

    agent.act(action, environment)
```
